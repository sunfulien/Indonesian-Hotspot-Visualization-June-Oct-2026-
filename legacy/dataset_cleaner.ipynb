```python
import pandas as pd
import os
```

CHAPTER 1: CONCATENATE DATASETS

```python
# Load the Excel file from the parent directory's 'dataset' folder
df1 = pd.read_excel("../dataset/sebaran-hotspot.xlsx")
nasa1_archive = pd.read_csv("../dataset/nasa/DL_FIRE_J1V-C2_823715/fire_archive_J1V-C2_823715.csv")
nasa1_nrt = pd.read_csv("../dataset/nasa/DL_FIRE_J1V-C2_823715/fire_nrt_J1V-C2_823715.csv")
nasa2_archive = pd.read_csv("../dataset/nasa/DL_FIRE_J2V-C2_823716/fire_archive_J2V-C2_823716.csv")
nasa2_nrt = pd.read_csv("../dataset/nasa/DL_FIRE_J2V-C2_823716/fire_nrt_J2V-C2_823716.csv")
nasa3_archive = pd.read_csv("../dataset/nasa/DL_FIRE_SV-C2_823717/fire_archive_SV-C2_823717.csv")
nasa3_nrt = pd.read_csv("../dataset/nasa/DL_FIRE_SV-C2_823717/fire_nrt_SV-C2_823717.csv")
```

```python
datasets = [nasa1_archive, nasa1_nrt, nasa2_archive, nasa2_nrt, nasa3_archive, nasa3_nrt]
subdatasets = [datasets[1], datasets[2], datasets[3], datasets[4] ]
for i, df in enumerate(subdatasets):
    print(f"Dataset {i} columns:", df.columns.tolist())
```

```python
datasets = [nasa1_archive, nasa1_nrt, nasa2_archive, nasa2_nrt, nasa3_archive, nasa3_nrt]
datasets_name = ["nasa1", "nasa2", "nasa3"]

j = 0

for i in range(0, len(datasets), 2):
    subdatasets = [datasets[i], datasets[i+1]]

    # Combine both dataframes
    combined = pd.concat(subdatasets, ignore_index=True)

    # Drop duplicates, keeping the first occurrence (which prioritizes 'archive' if you load it first)
    cleaned_data = combined.drop_duplicates(
        subset=['latitude', 'longitude', 'acq_date', 'acq_time'], 
        keep='first'
    )

    #save individual concatenated datasets
    os.makedirs("../dataset/nasa/concatenated", exist_ok=True)
    cleaned_data.to_csv(f"../dataset/nasa/concatenated/{datasets_name[j]}.csv", index=False)
    print(f"done {datasets_name[j]}")
    j += 1

```

CHAPTER 2: CLEANING DATASETS

```python
# Load the .csv file from concatenated folder
nasa1 = pd.read_csv("../dataset/nasa/concatenated/nasa1.csv")
nasa2 = pd.read_csv("../dataset/nasa/concatenated/nasa2.csv")
nasa3 = pd.read_csv("../dataset/nasa/concatenated/nasa3.csv")

nasa_datasets = [nasa1, nasa2, nasa3]
nasa1.head()
```

```python
# Check missing values count and percentage per column
for j, i in enumerate(nasa_datasets):
    print("checking:", datasets_name[j])
    missing_count = i.isna().sum()
    missing_pct = i.isna().mean() * 100

    summary_df = pd.DataFrame(
        {'Missing Count': missing_count, 'Missing Pct (%)': missing_pct}
    )
    print(summary_df.sort_values(by='Missing Pct (%)', ascending=False), "\n")
```

```python
n = nasa_datasets[0]
n.iloc[:4]
```

```python
nasa_cleaned = []
for j, i in enumerate(nasa_datasets): 
    print("CLEANING", datasets_name[j], "...")

    #check length
    print("Row Length prior:", len(i))

    #drop column "type"
    print("Removing Column 'type'...")
    #column before
    print("column before: ", i.columns.tolist())
    # Drop a single column
    df = i.drop(columns='type')
    #column after
    print("column after: ", df.columns.tolist())

    # remove confidence low (l)
    print("Filtering Confidence...")
    # removing row which column 'confidence' has cell value 'l' 
    df = df[df["confidence"] != "l"]

    # remove Nan
    print("Removing Nan...")
    df = df.dropna()
    #check aftermath length
    print("Row Length after Nan removal:", len(df))

    print("Row Length after confidence filtering: ", len(df))
    print("populating... \n")
    nasa_cleaned.append(df)

    #saving to .csv
    os.makedirs("../dataset/nasa/cleaned", exist_ok=True)
    df.to_csv(f"../dataset/nasa/cleaned/{datasets_name[j]}_cleaned.csv", index=False)

```

CHAPTER 3: CONCATENATE, SORT BY DATE --- TIME

```python
# Load the .csv file from cleaned folder
nasa1 = pd.read_csv("../dataset/nasa/cleaned/nasa1_cleaned.csv")
nasa2 = pd.read_csv("../dataset/nasa/cleaned/nasa2_cleaned.csv")
nasa3 = pd.read_csv("../dataset/nasa/cleaned/nasa3_cleaned.csv")

nasa_datasets = [nasa1, nasa2, nasa3]
nasa3.head()
```

```python
# 1. concetenate the three into one
combined = pd.concat(nasa_datasets, ignore_index=True)

# 2. Convert Date column to datetime type
combined['acq_date'] = pd.to_datetime(combined['acq_date'])

# 3. Sort by Date first, then by Time
combined_sorted = combined.sort_values(by=['acq_date', 'acq_time'], ascending=[True, True])

# 4. saving to .csv
os.makedirs("../dataset/nasa/final_dataset", exist_ok=True)
combined_sorted.to_csv(f"../dataset/nasa/final_dataset/nasa_final.csv", index=False)

len(combined_sorted)
```
